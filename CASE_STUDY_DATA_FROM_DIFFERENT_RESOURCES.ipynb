```python
# 1.Display the Top 5 countries by GDP.
cursor.execute('select country,GDP from country_statistics order by GDP DESC LIMIT 5 ')
rows=cursor.fetchall()
for row in rows:
  print(row)
```

```python
# 2.Find the average literacy rate.
cursor.execute('select AVG(literacy_rate) from country_statistics')
ro = cursor.fetchone()
print(ro[0])
```

```python
# 3.Display all countries having literacy rate above 90%.
cursor.execute('SELECT country, literacy_rate FROM country_statistics WHERE literacy_rate > 90.0')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# 4.Find the Top 5 countries with the highest internet penetration.
cursor.execute('''
SELECT country, "Internet Penetration (%)"
FROM country_statistics
ORDER BY "Internet Penetration (%)" DESC
LIMIT 5
''')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# 5. Calculate the total population.
cursor.execute('SELECT SUM(population) AS total_population FROM country_statistics')

rows = cursor.fetchone()

print(rows[0])
```

```python
# 6.Find the country with the highest GDP per capita.
cursor.execute('''SELECT country,"GDP Per Capita" FROM country_statistics ORDER BY "GDP Per Capita" DESC LIMIT 1''')
rows=cursor.fetchone()
print("Country:", rows[0])
print("GDP Per Capita:", rows[1])
```
